# 🧠 Handwritten Digit Recognition using Deep Neural Networks (MNIST)

## 📌 Project Overview

This project implements a **Deep Neural Network (DNN)** from scratch using **TensorFlow/Keras** to recognize handwritten digits from the **MNIST dataset**.

Rather than simply training a neural network, this project explores the complete machine learning workflow, including:

- Data preprocessing
- Model architecture design
- Overfitting analysis
- Model simplification
- L2 Regularization
- Hyperparameter tuning
- Performance evaluation
- Error analysis
- Interactive handwritten digit prediction

The goal was not only to build an accurate classifier but also to understand **why** each technique improves (or worsens) model performance.

---

# 📂 Dataset

**Dataset:** MNIST Handwritten Digits

- 70,000 grayscale images
- Image Size: **28 × 28 pixels**
- 10 classes (Digits 0–9)

Dataset split:

| Dataset | Samples |
|---------|---------:|
| Training Set | 60,000 |
| Cross Validation Set | 10,000 |
| Test Set | 10,000 |

---

# 🏗 Model Development

During this project, three different neural network architectures were developed and compared.

## 1️⃣ Complex Model

Architecture:

```
784
 ↓
Dense(250, ReLU)
 ↓
Dense(100, ReLU)
 ↓
Dense(10)
```

Observation:

- Extremely low training error
- Higher CV error
- Clear signs of overfitting

---

## 2️⃣ Simpler Model

Architecture:

```
784
 ↓
Dense(100, ReLU)
 ↓
Dense(35, ReLU)
 ↓
Dense(10)
```

Purpose:

Reduce model complexity to improve generalization.

---

## 3️⃣ Regularized Complex Model

Architecture:

```
784
 ↓
Dense(250, ReLU, L2)
 ↓
Dense(100, ReLU, L2)
 ↓
Dense(10)
```

L2 Regularization was introduced to reduce overfitting while preserving the expressive power of the larger network.

---

# 🔍 Hyperparameter Tuning

Instead of choosing a regularization constant manually, multiple values of λ (Lambda) were tested.

Example values:

```
0.000001
0.00001
0.0001
0.0002
0.0003
0.0004
0.0005
0.0006
0.0007
```

Each model was trained and evaluated on the Cross Validation set.

The value producing the lowest CV Classification Error was selected as the final model.

---

# 📈 Model Evaluation

Models were compared using:

- Classification Error
- Training Error
- Cross Validation Error
- Test Error

instead of relying only on loss values.

This provides a much better estimate of real-world performance.
